```python
import pandas as pd
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from bjontegaard import bd_rate, bd_psnr
```

```python

labels = ['INT4', 'INT8', 'INT12', 'INT16', 'FP32']
bd_rates = [254.0, 5.7, 0.2, 0.1, 0.0]
energys = [4.27, 8.55, 26.96, 45.33, 170.73]

plt.figure(figsize=(8, 4))
plt.scatter(energys, bd_rates, s=100)  # s controls point size
plt.grid(True, linestyle='--', alpha=0.6)

# Add labels next to each point
for x, y, label in zip(energys, bd_rates, labels):
    plt.text(x + 0.5, y + 1, label, fontsize=12)  # offset (+1,+1) avoids overlap

plt.xlabel('Energy Consumption (mJ)', fontsize=18)
plt.ylabel('BD-rate (%)', fontsize=18)
plt.tick_params(axis='both', labelsize=14)
plt.tight_layout()
plt.show()
```

```python

from turtle import color


labels = ['INT4', 'INT8', 'INT12', 'INT16', 'FP32']
bd_rates = [254.0, 5.7, 0.2, 0.1, 0.0]
energys = [4.27, 8.55, 26.96, 45.33, 170.73]

bd_rates_t1 = [266.9, 6.5, 0.0, 0.0, 0.0]

plt.figure(figsize=(8, 4))
plt.scatter(bd_rates, energys, s=100, label="Test configuration $t_{1}$", edgecolors='blue', facecolors='none', linewidths=2)  # s controls point size
plt.scatter(bd_rates_t1, energys, s=100, marker="x",  label="Test configuration $t_{2}$", color='#B22222', linewidths=2)  # s controls point size
plt.grid(True, linestyle='--', alpha=0.6)

# Add labels next to each point
for x, y, label in zip(bd_rates, energys, labels):
    plt.text(x + 1, y + 1, label, fontsize=14, color='black')  # offset (+1,+1) avoids overlap

plt.text(bd_rates[-1] + 25, energys[-1] - 15, '$\leftarrow$HIGH ENERGY\nCONSUMPTION', fontsize=14, color='red')  # offset (+1,+1) avoids overlap
plt.text(bd_rates[0] - 85, energys[0] + 15, 'PROHIBITVE CODING \nEFFICIENCY LOSS$\\searrow$', fontsize=14, color='red')  # offset (+1,+1) avoids overlap
plt.text(bd_rates[2] + 29, energys[2] - 24, '$\leftarrow$EFFICIENT\n   TRADE-OFF', fontsize=14, color='red')  # offset (+1,+1) avoids overlap


plt.ylabel('Energy per frame (mJ)', fontsize=18)
plt.ylim(-10, 200)

plt.xlabel('BD-Rate (%)', fontsize=18)
plt.ylim(-10, 250)
plt.tick_params(axis='both', labelsize=14)
plt.legend(fontsize=14)   # aumenta a legenda


theta = np.linspace(0, 2*np.pi, 500)
r = 16
x = r * np.cos(theta) + 12
y = r * np.sin(theta)*1.8 + 20
plt.plot(x, y, color='red', linestyle='--', linewidth=1)



plt.tight_layout()

plt.savefig("hw_analysis.pdf", dpi=600, bbox_inches="tight")   # formato PNG

plt.show()
```
